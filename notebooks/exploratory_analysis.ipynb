# Análise Exploratória - Estoque de Carros

Este notebook contém a análise exploratória dos dados de estoque da loja de carros.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
```

## 1. Carregamento dos Dados

```python
# Carregar dados
inventory_df = pd.read_csv('../datasets/car_inventory.csv')
sales_df = pd.read_csv('../datasets/sales_data.csv')

print(f"Estoque: {len(inventory_df)} registros")
print(f"Vendas: {len(sales_df)} registros")

inventory_df.head()
```

## 2. Análise do Estoque

```python
# Distribuição por marca
plt.figure(figsize=(12, 6))
inventory_df['marca'].value_counts().plot(kind='bar')
plt.title('Distribuição de Veículos por Marca')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Análise de preços
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
inventory_df['preco'].hist(bins=30)
plt.title('Distribuição de Preços')
plt.xlabel('Preço')

plt.subplot(1, 3, 2)
inventory_df.boxplot(column='preco', by='marca', ax=plt.gca())
plt.title('Preços por Marca')
plt.xticks(rotation=45)

plt.subplot(1, 3, 3)
inventory_df['status'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Status dos Veículos')

plt.tight_layout()
plt.show()
```

## 3. KPIs Principais

```python
# Calcular KPIs
kpis = {
    'Total de Veículos': len(inventory_df),
    'Valor Total do Estoque': f"R$ {inventory_df['preco'].sum():,.2f}",
    'Preço Médio': f"R$ {inventory_df['preco'].mean():,.2f}",
    'Veículos Disponíveis': len(inventory_df[inventory_df['status'] == 'Disponível']),
    'Taxa de Disponibilidade': f"{len(inventory_df[inventory_df['status'] == 'Disponível']) / len(inventory_df) * 100:.1f}%",
    'Marcas Ativas': inventory_df['marca'].nunique(),
    'Idade Média da Frota': f"{(2024 - inventory_df['ano']).mean():.1f} anos"
}

for key, value in kpis.items():
    print(f"{key}: {value}")
```

## 4. Análise de Vendas

```python
# Merge com dados de estoque
sales_with_inventory = sales_df.merge(inventory_df, left_on='car_id', right_on='id')

# Análise temporal de vendas
sales_with_inventory['data_venda'] = pd.to_datetime(sales_with_inventory['data_venda'])
vendas_mensais = sales_with_inventory.groupby(sales_with_inventory['data_venda'].dt.to_period('M')).size()

plt.figure(figsize=(12, 4))
vendas_mensais.plot(kind='line', marker='o')
plt.title('Vendas Mensais')
plt.xlabel('Mês')
plt.ylabel('Número de Vendas')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Top marcas vendidas
plt.figure(figsize=(10, 6))
sales_with_inventory['marca'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Marcas Mais Vendidas')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
